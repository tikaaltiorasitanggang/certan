```python
# Cara alternatif untuk membuat list
# range (awal, akhir-1, increment (harus integer))

data_range = range(0, 50, 2)
data_list = list(data_range)
print(data_list)
```

```python
# Membuat list dengan for
list = [i for i in range(0, 20, 2)]
print(list)
```

```python
# Membuat list dengan for dan if
list1 = [i for i in range(0, 20, 2) if i != 2]
print("list 1: ", list1)

list2 = [i for i in range(0, 20, 2) if i% 2 == 0]
print("list 2: ", list2)

list3 = [i for i in range(0, 20) if i % 2 != 0]
print("list 3: ", list3)
```
