```python
# Operasi elementwise (mengoperasikan setiap elemen yang berkorespondensi dengan elemen yang sama)
a = np.array([1, 2, 3, 4, 5])
b = np.array([6, 7, 8, 9, 10])

hasil1 = a + b
hasil2 = a - b
hasil3 = a * b
hasil4 = a / b
hasil5 = a ** 2

print("hasil 1 =", hasil1)
print("hasil 2 =", hasil2)
print("hasil 3 =", hasil3)
print("hasil 4 =", hasil4)
print("hasil 5 =", hasil5)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d

# =========================
# Plotting dengan pyplot style
# =========================

x = np.linspace(-5, 5, 100)

plt.plot(x, x, label='linear')
plt.plot(x, x**2, label='quadratic')
plt.plot(x, x**3, label='cubic')

plt.xlabel('x label')
plt.ylabel('y label')
plt.title('Simple Plot')
plt.legend()
plt.show()


# =========================
# Plotting 3D dengan Matplotlib
# =========================

fig = plt.figure()
ax = plt.axes(projection='3d')


# Fungsi:
# f(x, y) = e^(-(x^2 + y^2)) sin(sqrt(x^2 + y^2))

def f(x, y):
    return np.exp(-x**2 - y**2) * np.sin(np.sqrt(x**2 + y**2))


x = np.linspace(-np.pi, np.pi, 30)
y = np.linspace(-np.pi, np.pi, 30)

X, Y = np.meshgrid(x, y)
Z = f(X, Y)


fig = plt.figure()
ax = plt.axes(projection='3d')

ax.contour3D(X, Y, Z, 50, cmap='binary')

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')

plt.show()
```

```python
# ==========================================================
# TUGAS PRAKTIKUM TEKNOLOGI KECERDASAN BUATAN - MINGGU 1
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d


# ==========================================================
# SOAL 1
# Buatlah sebuah Matriks berukuran 5x5 dengan elemen random.
# Kemudian dari elemen yang dihasilkan perlihatkan nilai
# maksimum, nilai minimum, dan nilai rata-rata dari suatu
# elemen matriks yang dikonstruksi.
# ==========================================================

# Membuat matriks 5x5 dengan elemen random (integer 1-100)
matriks = np.random.randint(1, 100, size=(5, 5))

print("=== SOAL 1 ===")
print("Matriks 5x5 random:")
print(matriks)
print()
print("Nilai maksimum :", matriks.max())
print("Nilai minimum  :", matriks.min())
print("Nilai rata-rata:", matriks.mean())
print()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================================
# SOAL 2
# Buatlah grafik 2 dimensi dari persamaan berikut:
# x^6 + 3x^4y^2 - 3x^4 + 3x^2y^4 - x^2y^3 - 6x^2y^2
#   + 3x^2 + y^6 - 3y^4 + 3y^2 = 1
# ==========================================================

def F_soal2(x, y):
    return (x**6 + 3*x**4*y**2 - 3*x**4 
            + 3*x**2*y**4 - x**2*y**3 - 6*x**2*y**2
            + 3*x**2 + y**6 - 3*y**4 + 3*y**2)

x2 = np.linspace(-2.5, 2.5, 500)
y2 = np.linspace(-2.5, 2.5, 500)

X2, Y2 = np.meshgrid(x2, y2)

Z2 = F_soal2(X2, Y2)

fig2, ax2 = plt.subplots(figsize=(6, 6))

ax2.contour(
    X2, Y2, Z2,
    levels=[1],
    colors='blue'
)

ax2.set_xlabel('x')
ax2.set_ylabel('y')

ax2.set_title(
    r'Soal 2: $x^6+3x^4y^2-3x^4+3x^2y^4-x^2y^3-6x^2y^2+3x^2+y^6-3y^4+3y^2=1$',
    fontsize=8
)

ax2.axhline(0, color='gray', linewidth=0.5)
ax2.axvline(0, color='gray', linewidth=0.5)

ax2.set_aspect('equal')

plt.show()
```

```python
# ==========================================================
# SOAL 3
# Buatlah grafik 3 dimensi dari persamaan berikut:
# z = (5y/4 - |x|)^2 + x^2
# ==========================================================

def f_soal3(x, y):
    return (5*y/4 - np.abs(x))*2 + x*2

x3 = np.linspace(-3, 3, 100)
y3 = np.linspace(-3, 3, 100)
X3, Y3 = np.meshgrid(x3, y3)
Z3 = f_soal3(X3, Y3)

fig3 = plt.figure(figsize=(8, 6))
ax3 = plt.axes(projection='3d')
ax3.plot_surface(X3, Y3, Z3, cmap='viridis')
ax3.set_xlabel('x')
ax3.set_ylabel('y')
ax3.set_zlabel('z')
ax3.set_title(r'Soal 3: $z=\left(\frac{5y}{4}-|x|\right)^2+x^2$')
plt.show()
```
