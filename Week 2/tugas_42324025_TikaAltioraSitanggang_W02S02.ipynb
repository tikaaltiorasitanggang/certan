```python
import pandas as pd
path = "GlobalWeatherRepository.csv"
df = pd.read_csv(path)

df
```

```python
df[
    [
        'temperature_fahrenheit',
        'pressure_mb',
        'air_quality_Sulphur_dioxide'
    ]
].corr()
```

### Tugas 1 Analisis Korelasi

Berdasarkan hasil analisis korelasi, diperoleh:

- Temperature Fahrenheit dengan Pressure mb memiliki korelasi sebesar **-0.486833**, yang menunjukkan hubungan negatif sedang. Artinya, ketika temperature_fahrenheit meningkat, pressure_mb cenderung menurun.
- Temperature Fahrenheit dengan Air Quality Sulphur Dioxide memiliki korelasi sebesar **-0.023152**, yang menunjukkan hubungan negatif yang sangat lemah. Artinya, hampir tidak terdapat hubungan linear antara kedua variabel tersebut.
- Pressure mb dengan Air Quality Sulphur Dioxide memiliki korelasi sebesar **0.018620**, yang menunjukkan hubungan positif yang sangat lemah. Artinya, hampir tidak terdapat hubungan linear antara kedua variabel tersebut.

Dapat disimpulkan bahwa hubungan yang paling kuat di antara ketiga variabel tersebut adalah hubungan antara temperature_fahrenheit dan pressure_mb dengan nilai korelasi **-0.486833**.

```python
import pandas as pd

path_customer = "shopping_trends.csv"
df_customer = pd.read_csv(path_customer)

df_customer
```

```python
previous_purchases = df_customer['Previous Purchases']

print("Mean :", previous_purchases.mean())
print("Std  :", previous_purchases.std())
print("Min  :", previous_purchases.min())
print("Q1   :", previous_purchases.quantile(0.25))
print("Q2   :", previous_purchases.quantile(0.50))
print("Q3   :", previous_purchases.quantile(0.75))
print("Max  :", previous_purchases.max())
```

### Tugas 2a Analisis Previous Purchases

Berdasarkan hasil analisis statistik deskriptif pada data **Previous Purchases**, diperoleh:

- **Mean** = 25.351538
- **Std** = 14.447125
- **Min** = 1
- **Q1** = 13
- **Q2 (Median)** = 25
- **Q3** = 38
- **Max** = 50

Nilai mean sebesar **25.351538** menunjukkan bahwa rata-rata jumlah pembelian sebelumnya adalah sekitar 25 pembelian. Nilai standar deviasi sebesar **14.447125** menunjukkan adanya penyebaran data yang cukup besar dari nilai rata-ratanya.

Nilai minimum adalah **1** dan maksimum adalah **50**. Berdasarkan nilai kuartil, 25% data memiliki Previous Purchases kurang dari atau sama dengan **13**, 50% data kurang dari atau sama dengan **25**, dan 75% data kurang dari atau sama dengan **38**.
